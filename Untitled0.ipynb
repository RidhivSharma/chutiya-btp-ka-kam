```python
# ============================================================
# Cell 13: Cleaning
# ============================================================

clean_data = {}

for name, df in data.items():

    temp = df.copy()

    # Replace infinite values with NaN
    temp[FEATURES] = temp[FEATURES].replace(
        [np.inf, -np.inf],
        np.nan
    )

    # Interpolate internal missing values
    temp[FEATURES] = temp[FEATURES].interpolate(
        method="linear",
        limit_direction="both"
    )

    # Remove rows that still contain invalid values
    temp = temp.dropna(subset=FEATURES)

    # Reset index
    temp = temp.reset_index(drop=True)

    clean_data[name] = temp

print("Cleaning completed.")

for name in clean_data:
    print(f"{name}: {len(clean_data[name]):,} rows")
```

```python
# ============================================================
# Cell 14: Sampling interval verification
# ============================================================

sampling_results = []

for name, df in clean_data.items():

    time_values = df["Time"].to_numpy()

    dt = np.diff(time_values)

    # Remove zero/invalid differences
    dt = dt[np.isfinite(dt)]
    dt = dt[dt > 0]

    median_dt = np.median(dt)
    mean_dt = np.mean(dt)

    sampling_results.append({
        "Dataset": name,
        "Median dt": median_dt,
        "Mean dt": mean_dt,
        "Estimated Fs": 1 / median_dt
    })

sampling_df = pd.DataFrame(sampling_results)

display(sampling_df)
```

```python
# ============================================================
# Cell 15: Downsampling
# ============================================================

DOWNSAMPLE_FACTOR = 10

downsampled_data = {}

for name, df in clean_data.items():

    # Take every 10th sample
    temp = df.iloc[::DOWNSAMPLE_FACTOR].copy()

    temp = temp.reset_index(drop=True)

    downsampled_data[name] = temp

print("Downsampling completed.")

for name in downsampled_data:
    print(
        f"{name}: "
        f"{len(clean_data[name]):,} → "
        f"{len(downsampled_data[name]):,} rows"
    )
```

```python
# ============================================================
# Cell 22: Temporal window generation
# ============================================================

def create_windows(X, window_size=100, stride=10):

    windows = []

    for start in range(
        0,
        len(X) - window_size + 1,
        stride
    ):

        window = X[
            start:start + window_size
        ]

        windows.append(window)

    return np.asarray(
        windows,
        dtype=np.float32
    )


# Healthy MPPT
X_F0M_windows = create_windows(
    X_data["F0M"],
    WINDOW_SIZE,
    STRIDE
)

# Healthy IPPT
X_F0L_windows = create_windows(
    X_data["F0L"],
    WINDOW_SIZE,
    STRIDE
)

print("F0M window shape:", X_F0M_windows.shape)
print("F0L window shape:", X_F0L_windows.shape)
```

```python
# ============================================================
# Cell 28: Deep learning imports
# ============================================================

import tensorflow as tf

from tensorflow.keras import Model
from tensorflow.keras.layers import (
    Input,
    Dense,
    Conv1D,
    BatchNormalization,
    Activation,
    Dropout,
    Add,
    Bidirectional,
    LSTM,
    RepeatVector,
    TimeDistributed
)

from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)

print("TensorFlow version:", tf.__version__)

print(
    "GPU available:",
    tf.config.list_physical_devices("GPU")
)
```

```python
# ============================================================
# Cell 29: TCN residual block
# ============================================================

def tcn_residual_block(
    x,
    filters,
    kernel_size=3,
    dilation_rate=1,
    dropout_rate=0.1
):

    shortcut = x

    # First dilated convolution
    y = Conv1D(
        filters=filters,
        kernel_size=kernel_size,
        padding="same",
        dilation_rate=dilation_rate
    )(x)

    y = BatchNormalization()(y)
    y = Activation("relu")(y)
    y = Dropout(dropout_rate)(y)

    # Second dilated convolution
    y = Conv1D(
        filters=filters,
        kernel_size=kernel_size,
        padding="same",
        dilation_rate=dilation_rate
    )(y)

    y = BatchNormalization()(y)
    y = Activation("relu")(y)
    y = Dropout(dropout_rate)(y)

    # Match shortcut dimensions if required
    if shortcut.shape[-1] != filters:
        shortcut = Conv1D(
            filters=filters,
            kernel_size=1,
            padding="same"
        )(shortcut)

    # Residual connection
    y = Add()([shortcut, y])
    y = Activation("relu")(y)

    return y
```

```python
# ============================================================
# Cell 30: Proposed TCN-BiLSTM Autoencoder
# ============================================================

def build_tcn_bilstm_autoencoder(
    timesteps=100,
    n_features=13
):

    inputs = Input(
        shape=(timesteps, n_features),
        name="input_sequence"
    )

    # --------------------------------------------------------
    # TCN Encoder
    # --------------------------------------------------------

    x = tcn_residual_block(
        inputs,
        filters=32,
        kernel_size=3,
        dilation_rate=1
    )

    x = tcn_residual_block(
        x,
        filters=32,
        kernel_size=3,
        dilation_rate=2
    )

    x = tcn_residual_block(
        x,
        filters=64,
        kernel_size=3,
        dilation_rate=4
    )

    # --------------------------------------------------------
    # Bi-LSTM Encoder
    # --------------------------------------------------------

    x = Bidirectional(
        LSTM(
            64,
            return_sequences=True
        )
    )(x)

    # Latent representation
    latent = Bidirectional(
        LSTM(
            32,
            return_sequences=False
        ),
        name="latent_encoder"
    )(x)

    # --------------------------------------------------------
    # Decoder
    # --------------------------------------------------------

    x = RepeatVector(timesteps)(latent)

    x = Bidirectional(
        LSTM(
            32,
            return_sequences=True
        )
    )(x)

    x = Bidirectional(
        LSTM(
            64,
            return_sequences=True
        )
    )(x)

    # --------------------------------------------------------
    # TCN Decoder
    # --------------------------------------------------------

    x = tcn_residual_block(
        x,
        filters=64,
        kernel_size=3,
        dilation_rate=4
    )

    x = tcn_residual_block(
        x,
        filters=32,
        kernel_size=3,
        dilation_rate=2
    )

    # --------------------------------------------------------
    # Reconstruction
    # --------------------------------------------------------

    outputs = TimeDistributed(
        Dense(n_features)
    )(x)

    model = Model(
        inputs,
        outputs,
        name="TCN_BiLSTM_Autoencoder"
    )

    return model
```

```python
# ============================================================
# Cell 31: MPPT model
# ============================================================

mppt_model = build_tcn_bilstm_autoencoder(
    timesteps=WINDOW_SIZE,
    n_features=len(MODEL_FEATURES)
)

mppt_model.compile(
    optimizer=Adam(learning_rate=1e-3),
    loss="mse"
)

mppt_model.summary()
```

```python
# ============================================================
# Cell 54: Balanced overall evaluation
# ============================================================

def balanced_autoencoder_evaluation(
    healthy_errors,
    fault_errors_dict,
    threshold,
    random_state=42
):

    rng = np.random.default_rng(random_state)

    # Combine all fault errors
    fault_errors = np.concatenate(
        list(fault_errors_dict.values())
    )

    # Equal number of healthy and faulty windows
    n = min(
        len(healthy_errors),
        len(fault_errors)
    )

    # Randomly select healthy and fault windows
    healthy_idx = rng.choice(
        len(healthy_errors),
        size=n,
        replace=False
    )

    fault_idx = rng.choice(
        len(fault_errors),
        size=n,
        replace=False
    )

    selected_healthy = healthy_errors[healthy_idx]
    selected_faults = fault_errors[fault_idx]

    # Construct test set
    errors = np.concatenate([
        selected_healthy,
        selected_faults
    ])

    y_true = np.concatenate([
        np.zeros(n, dtype=int),
        np.ones(n, dtype=int)
    ])

    # Predictions
    y_pred = (
        errors > threshold
    ).astype(int)

    # Shuffle
    indices = rng.permutation(len(errors))

    return (
        y_true[indices],
        y_pred[indices],
        errors[indices]
    )


# MPPT
mppt_bal_y_true, mppt_bal_y_pred, mppt_bal_errors = (
    balanced_autoencoder_evaluation(
        mppt_healthy_errors,
        mppt_fault_errors,
        mppt_threshold
    )
)


# IPPT
ippt_bal_y_true, ippt_bal_y_pred, ippt_bal_errors = (
    balanced_autoencoder_evaluation(
        ippt_healthy_errors,
        ippt_fault_errors,
        ippt_threshold
    )
)


print("MPPT balanced evaluation:")
print("Healthy:", np.sum(mppt_bal_y_true == 0))
print("Fault  :", np.sum(mppt_bal_y_true == 1))

print("\nIPPT balanced evaluation:")
print("Healthy:", np.sum(ippt_bal_y_true == 0))
print("Fault  :", np.sum(ippt_bal_y_true == 1))
```

```python
# ============================================================
# Cell 55: Final balanced metrics
# ============================================================

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

def complete_metrics(y_true, y_pred):

    cm = confusion_matrix(
        y_true,
        y_pred
    )

    tn, fp, fn, tp = cm.ravel()

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else 0
    )

    fpr = (
        fp / (fp + tn)
        if (fp + tn) > 0
        else 0
    )

    return {
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "Specificity": specificity,
        "FPR": fpr,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    }


mppt_final_metrics = complete_metrics(
    mppt_bal_y_true,
    mppt_bal_y_pred
)

ippt_final_metrics = complete_metrics(
    ippt_bal_y_true,
    ippt_bal_y_pred
)


print("=" * 65)
print("PROPOSED MODEL — BALANCED EVALUATION")
print("=" * 65)

print("\nMPPT")
for key, value in mppt_final_metrics.items():
    print(f"{key:15s}: {value:.4f}" if isinstance(value, float)
          else f"{key:15s}: {value}")

print("\nIPPT")
for key, value in ippt_final_metrics.items():
    print(f"{key:15s}: {value:.4f}" if isinstance(value, float)
          else f"{key:15s}: {value}")
```

```python
# ============================================================
# Cell 58 — Reconstruction error calculation
# ============================================================

def calculate_reconstruction_errors(model, X):
    """
    Calculate:
      1. Overall reconstruction error per window
      2. Feature-wise reconstruction error per window

    X shape:
        (n_windows, timesteps, n_features)
    """

    X_pred = model.predict(X, verbose=0)

    # Squared reconstruction error
    sq_error = np.square(X - X_pred)

    # Overall error:
    # mean over time and features
    window_error = np.mean(sq_error, axis=(1, 2))

    # Feature-wise error:
    # mean over time only
    feature_error = np.mean(sq_error, axis=1)

    return window_error, feature_error, X_pred


# ------------------------------------------------------------
# Healthy IPPT validation
# ------------------------------------------------------------

healthy_errors, healthy_feature_errors, healthy_pred = \
    calculate_reconstruction_errors(
        MODEL,
        X_ippt_val_seq
    )

print("Healthy reconstruction errors calculated.")

print("Healthy error shape:")
print(healthy_errors.shape)

print("\nHealthy feature-error shape:")
print(healthy_feature_errors.shape)

print("\nHealthy statistics:")
print(pd.Series(healthy_errors).describe())
```

```python
# ============================================================
# SECTION 5B — FEATURE-NORMALIZED ANOMALY SCORE
# Cell 69: Feature-wise reconstruction error
# ============================================================

def feature_reconstruction_error(model, X):
    """
    Returns reconstruction error separately for each feature.

    Input:
        X = (n_windows, timesteps, n_features)

    Output:
        feature_errors = (n_windows, n_features)
    """

    reconstructed = model.predict(
        X,
        batch_size=64,
        verbose=0
    )

    squared_error = np.square(
        X - reconstructed
    )

    # Average only over temporal dimension
    feature_errors = np.mean(
        squared_error,
        axis=1
    )

    return feature_errors


# Existing trained IPPT model
ippt_feature_errors_v2 = {}

# Healthy validation
ippt_feature_errors_v2["F0L"] = feature_reconstruction_error(
    ippt_model,
    X_ippt_val_seq
)

# Faults
for i in range(1, 8):

    fault_name = f"F{i}L"

    ippt_feature_errors_v2[fault_name] = \
        feature_reconstruction_error(
            ippt_model,
            windowed_data[fault_name]
        )


print("Feature-wise reconstruction errors calculated.")

for name, errors in ippt_feature_errors_v2.items():

    print(
        f"{name}: {errors.shape}"
    )
```

```python
# ============================================================
# Cell 70 — Healthy feature normalization
# ============================================================

healthy_feature_errors = ippt_feature_errors_v2["F0L"]

# Healthy reference statistics
healthy_feature_mean = np.mean(
    healthy_feature_errors,
    axis=0
)

healthy_feature_std = np.std(
    healthy_feature_errors,
    axis=0
)

# Prevent division by zero
EPS = 1e-8

healthy_feature_std = np.maximum(
    healthy_feature_std,
    EPS
)

healthy_feature_reference = pd.DataFrame({
    "Feature": FEATURE_NAMES,
    "Healthy Mean Error": healthy_feature_mean,
    "Healthy Std Error": healthy_feature_std
})

display(
    healthy_feature_reference.round(6)
)
```

```python
# ============================================================
# Cell 71 — Feature-normalized anomaly score
# ============================================================

TOP_K = 3


def normalized_feature_score(
    feature_errors,
    healthy_mean,
    healthy_std,
    top_k=3
):
    """
    Convert feature-wise reconstruction errors into
    a normalized anomaly score.

    For each window:
        z_f = (error_f - healthy_mean_f) / healthy_std_f

    Only positive deviations are considered.

    Final score = mean of top-k positive feature deviations.
    """

    z_scores = (
        feature_errors - healthy_mean
    ) / healthy_std

    # Only abnormal positive deviations matter
    z_scores = np.maximum(
        z_scores,
        0
    )

    # Sort feature scores per window
    sorted_scores = np.sort(
        z_scores,
        axis=1
    )[:, ::-1]

    k = min(
        top_k,
        sorted_scores.shape[1]
    )

    score = np.mean(
        sorted_scores[:, :k],
        axis=1
    )

    return score


ippt_normalized_scores = {}

for condition, feature_errors in ippt_feature_errors_v2.items():

    ippt_normalized_scores[condition] = \
        normalized_feature_score(
            feature_errors,
            healthy_feature_mean,
            healthy_feature_std,
            TOP_K
        )


print("Feature-normalized scores calculated.")

for condition, scores in ippt_normalized_scores.items():

    print(
        f"{condition}: "
        f"mean={np.mean(scores):.4f}, "
        f"median={np.median(scores):.4f}"
    )
```

```python
# ============================================================
# Cell 72 — Adaptive threshold for normalized score
# ============================================================

IPPT_NORM_THRESHOLD = np.percentile(
    ippt_normalized_scores["F0L"],
    99
)

print("=" * 60)
print("FEATURE-NORMALIZED IPPT THRESHOLD")
print("=" * 60)

print(
    f"Threshold (99th percentile): "
    f"{IPPT_NORM_THRESHOLD:.6f}"
)

print(
    "\nHealthy false-positive rate:",
    np.mean(
        ippt_normalized_scores["F0L"]
        > IPPT_NORM_THRESHOLD
    ) * 100,
    "%"
)
```

```python
# ============================================================
# Cell 73 — Fault-wise detection
# ============================================================

normalized_fault_results = []

for i in range(1, 8):

    fault_name = f"F{i}L"

    scores = ippt_normalized_scores[fault_name]

    detected = np.sum(
        scores > IPPT_NORM_THRESHOLD
    )

    total = len(scores)

    detection = (
        detected / total * 100
    )

    normalized_fault_results.append({
        "Fault": fault_name,
        "Total": total,
        "Detected": detected,
        "Missed": total - detected,
        "Detection (%)": detection
    })


normalized_fault_df = pd.DataFrame(
    normalized_fault_results
)

display(
    normalized_fault_df.round(3)
)
```

```python
# ============================================================
# SECTION 6 — TEMPORAL DELAY ROBUSTNESS
# Cell 77: Delay simulation
# ============================================================

# Final improved IPPT threshold
FINAL_IPPT_THRESHOLD = IPPT_NORM_THRESHOLD

# Delay values in milliseconds
DELAY_MS = [
    0,
    10,
    50,
    100,
    500,
    1000
]

TARGET_FS = 1000  # Hz


def apply_temporal_delay(X, delay_ms, fs=1000):
    """
    Simulate delayed/stale measurements by shifting the
    temporal sequence forward by the requested delay.

    The first samples are padded using the earliest available
    sample to avoid introducing artificial zeros.
    """

    delay_samples = int(
        delay_ms * fs / 1000
    )

    if delay_samples == 0:
        return X.copy()

    if delay_samples >= len(X):
        delay_samples = len(X) - 1

    X_delayed = np.empty_like(X)

    X_delayed[:delay_samples] = X[0]

    X_delayed[delay_samples:] = (
        X[:-delay_samples]
    )

    return X_delayed


print("Delay robustness configuration")
print("=" * 60)

for delay in DELAY_MS:
    print(
        f"{delay:5d} ms "
        f"-> {int(delay * TARGET_FS / 1000):6d} samples"
    )
```

```python
# ============================================================
# Cell 78: Generate delayed IPPT windows
# ============================================================

delayed_ippt_data = {}

conditions = [
    "F0L",
    "F1L",
    "F2L",
    "F3L",
    "F4L",
    "F5L",
    "F6L",
    "F7L"
]

for delay in DELAY_MS:

    delayed_ippt_data[delay] = {}

    for condition in conditions:

        X_original = (
            X_ippt_val_seq
            if condition == "F0L"
            else windowed_data[condition]
        )

        delayed_ippt_data[delay][condition] = \
            apply_temporal_delay(
                X_original,
                delay,
                TARGET_FS
            )

    print(
        f"{delay} ms delay generated."
    )
```

```python
# ============================================================
# Cell 79: Calculate delayed anomaly scores
# ============================================================

def calculate_normalized_scores_for_X(X):

    feature_errors = feature_reconstruction_error(
        ippt_model,
        X
    )

    scores = normalized_feature_score(
        feature_errors,
        healthy_feature_mean,
        healthy_feature_std,
        TOP_K
    )

    return scores


delayed_scores = {}

for delay in DELAY_MS:

    delayed_scores[delay] = {}

    for condition in conditions:

        delayed_scores[delay][condition] = \
            calculate_normalized_scores_for_X(
                delayed_ippt_data[delay][condition]
            )

    print(
        f"{delay} ms: scores calculated"
    )
```

```python
# ============================================================
# Cell 80: Fault detection vs delay
# ============================================================

delay_fault_results = []

for delay in DELAY_MS:

    row = {
        "Delay (ms)": delay
    }

    for i in range(1, 8):

        condition = f"F{i}L"

        scores = delayed_scores[delay][condition]

        detection = np.mean(
            scores > FINAL_IPPT_THRESHOLD
        ) * 100

        row[f"F{i} Detection (%)"] = detection

    delay_fault_results.append(row)


delay_fault_df = pd.DataFrame(
    delay_fault_results
)

display(
    delay_fault_df.round(2)
)
```

```python
# ============================================================
# Cell 81: Balanced performance under delay
# ============================================================

delay_metrics = []

rng = np.random.default_rng(42)

for delay in DELAY_MS:

    healthy_scores = delayed_scores[delay]["F0L"]

    fault_scores = np.concatenate([
        delayed_scores[delay][f"F{i}L"]
        for i in range(1, 8)
    ])

    n = min(
        len(healthy_scores),
        len(fault_scores)
    )

    # Same balanced protocol
    healthy_idx = rng.choice(
        len(healthy_scores),
        size=n,
        replace=False
    )

    fault_idx = rng.choice(
        len(fault_scores),
        size=n,
        replace=False
    )

    selected_healthy = healthy_scores[
        healthy_idx
    ]

    selected_faults = fault_scores[
        fault_idx
    ]

    scores = np.concatenate([
        selected_healthy,
        selected_faults
    ])

    y_true = np.concatenate([
        np.zeros(n, dtype=int),
        np.ones(n, dtype=int)
    ])

    y_pred = (
        scores > FINAL_IPPT_THRESHOLD
    ).astype(int)

    # Shuffle
    indices = rng.permutation(len(scores))

    y_true = y_true[indices]
    y_pred = y_pred[indices]

    metrics = complete_metrics(
        y_true,
        y_pred
    )

    delay_metrics.append({
        "Delay (ms)": delay,
        "Accuracy": metrics["Accuracy"],
        "Precision": metrics["Precision"],
        "Recall": metrics["Recall"],
        "F1": metrics["F1"],
        "Specificity": metrics["Specificity"],
        "FPR": metrics["FPR"],
        "FN": metrics["FN"],
        "FP": metrics["FP"]
    })


delay_metrics_df = pd.DataFrame(
    delay_metrics
)

display(
    delay_metrics_df.round(4)
)
```
